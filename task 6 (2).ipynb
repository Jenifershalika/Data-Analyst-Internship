```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("retail_sales_dataset.csv")
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
df.columns.tolist()
```

```python
plt.figure(figsize=(8,5))
df.groupby("Product Category")["Total Amount"].sum().plot(kind="bar")
plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Amount")

plt.savefig("bar_chart.png", bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8,5))

df["Date"] = pd.to_datetime(df["Date"])
df.groupby("Date")["Total Amount"].sum().plot(kind="line")

plt.title("Sales Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Total Amount")

plt.savefig("line_chart.png", bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Total Amount"], bins=10)

plt.title("Distribution of Total Amount")
plt.xlabel("Total Amount")
plt.ylabel("Frequency")

plt.savefig("histogram.png", bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Quantity"], df["Total Amount"])

plt.title("Quantity vs Total Amount")
plt.xlabel("Quantity")
plt.ylabel("Total Amount")

plt.savefig("scatter_plot.png", bbox_inches="tight")
plt.show()
```

## Insights

1. The highest total sales come from specific product categories, showing which categories drive most revenue.
2. Sales over time show clear variations, indicating seasonal or demand-based trends.
3. There is a positive relationship between quantity sold and total amount, meaning higher quantity generally leads to higher sales value.
